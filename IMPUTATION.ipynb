```python
##modified
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer, KNNImputer
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

def load_methylation_data(file_path):
    try:
        data = pd.read_csv(file_path,header=None)
        print(f"Successfully loaded data with shape: {data.shape}")
        return data
    except Exception as e:
        print(f"Error loading data: {str(e)}")
        return None

def impute_methylation_data(data, k_neighbors=15, strategy='mean'):

    """
    Impute only missing values using KNN while preserving existing values.
    """
    data_copy = data.copy()

    # Calculate missing value statistics before imputation
    total_values = data.size
    missing_values = data.isna().sum().sum()
    missing_percentage = (missing_values / total_values) * 100

    stats = {
        'total_values': total_values,
        'missing_values_before': missing_values,
        'missing_percentage_before': missing_percentage,
    }

    # Apply KNN imputation
    imputer = KNNImputer(n_neighbors=k_neighbors)
    imputed_data = imputer.fit_transform(data_copy)

    # Convert back to DataFrame
    imputed_data = pd.DataFrame(imputed_data, index=data_copy.index, columns=data_copy.columns)


    stats['k_neighbors'] = k_neighbors
    stats['missing_values_after'] = imputed_data.isna().sum().sum()
    stats['missing_percentage_after'] = (stats['missing_values_after'] / total_values) * 100
    stats['mean_difference'] = abs(data_copy.mean().mean() - imputed_data.mean().mean())
    stats['std_difference'] = abs(data_copy.std().mean() - imputed_data.std().mean())

    return imputed_data, stats

def save_results(imputed_data, stats, validation, output_dir):

    # Save imputed data
    imputed_data.to_csv(output_path)

from google.colab import files

def main():
    """
    Main function to run the complete methylation data imputation pipeline.
    """

    # Upload input file
    print("Please upload your methylation CSV file (TCGA_DNAm_processed.csv)...")
    uploaded = files.upload()
    file_path = list(uploaded.keys())[0]

    # Load data
    data = load_methylation_data(file_path)
    if data is None:
        return

    # Perform imputation
    print("\nPerforming imputation...")
    imputed_data, stats = impute_methylation_data(data)

    # Save results locally
    output_path = "DNAm_processed_imputed.csv"
    imputed_data.to_csv(output_path, index=False, header=False)

    # Download the file
    files.download(output_path)

    print(f"\nAnalysis complete! Results saved and ready to download: {output_path}")

    # Print summary
    print("\nSummary:")
    print(f"Total samples: {len(data)}")
    print(f"Total CpG sites: {len(data.columns)}")
    print(f"Missing values before imputation: {stats['missing_values_before']}")
    print(f"Missing values after imputation: {stats['missing_values_after']}")


if __name__ == "__main__":
    main()
```
